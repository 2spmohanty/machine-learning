### Summary of the Squared Error Cost Function J(w,b)

* **Overall Fit vs. Single Error:** $J(w,b)$ measures the average of the squared residuals across all training examples, capturing overall model performance rather than a single prediction's error.
* **Eliminating Direction:** Squaring forces both negative and positive residuals to become positive, preventing opposite errors from canceling out and hiding poor fit.
* **Punishing Large Outliers:** Squaring amplifies larger errors disproportionately, penalising significant misses much harder than minor miscalculations.


### Why Gradient Descent Works Universally

The gradient descent formula does not need any complex logic or "if/else" conditions to decide which way to move. The subtraction step itself handles everything perfectly based on the slope of the curve.

* **When the slope is negative:** 
  * The derivative value is negative.
  * Subtracting a negative number turns into addition ($w - (-\text{value}) = w + \text{value}$).
  * This automatically **increases** the weight ($w$) to move it forward.

* **When the slope is positive:** 
  * The derivative value is positive.
  * Subtracting a positive number reduces the value ($w - \text{value}$).
  * This automatically **decreases** the weight ($w$) to move it backward.

Because subtraction inherently reverses the direction of the slope, the single universal formula works perfectly in all directions:

Weight = Weight - (Learning Rate * Slope)
